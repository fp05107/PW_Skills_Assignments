## **1. What is Matplotlib and what are its key features?**  
**Answer:**  
Matplotlib is a powerful Python library for creating static, animated, and interactive visualizations. It is highly customizable and widely used for data visualization in Python.

### **Key Features of Matplotlib:**  
- **Versatility:** Supports various plot types (line, bar, scatter, histogram, etc.).  
- **Customization:** Allows fine-tuning of plots (colors, labels, legends, axes, etc.).  
- **Backend Support:** Works with different backends (e.g., `Tkinter`, `Qt`, `Jupyter Notebook`).  
- **Integration:** Works well with NumPy, Pandas, and Seaborn.  
- **Publication-Quality Plots:** High-resolution output in multiple formats (PNG, PDF, SVG).  
- **Subplots & Layouts:** Supports multiple plots in a single figure.  
- **Annotations & Text:** Enables adding text, arrows, and annotations.

## **2. How do you create a basic line plot using Matplotlib?**  
**Answer:**  
```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

# Create a line plot
plt.plot(x, y, label="Linear Growth", color="blue", linestyle="-", marker="o")

# Add labels and title
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Basic Line Plot")
plt.legend()

# Display plot
plt.show()
```
**Explanation:**  
- `plt.plot()` creates the line plot.  
- Customizations: `color`, `linestyle`, `marker`, `label`.  
- `plt.xlabel()`, `plt.ylabel()`, `plt.title()` add labels.  
- `plt.legend()` shows the legend.

## **3. Explain how to customize plot labels, titles, and legends in Matplotlib.**  
**Answer:**  
### **Customizing Labels & Titles:**  
```python
plt.xlabel("X-axis Label", fontsize=12, color="red", fontweight="bold")
plt.ylabel("Y-axis Label", fontsize=12, color="green")
plt.title("Customized Title", fontsize=16, color="blue", pad=20)
```
- `fontsize`: Controls text size.  
- `color`: Sets text color.  
- `fontweight`: Makes text bold.  
- `pad`: Adds padding to the title.  

### **Customizing Legends:**  
```python
plt.plot(x, y, label="Data Line")
plt.legend(loc="upper right", title="Legend", fontsize=10, shadow=True)
```
- `loc`: Position (`upper right`, `lower left`, etc.).  
- `title`: Adds a legend title.  
- `shadow`: Adds a shadow effect.

## **4. What are subplots in Matplotlib, and how do you use them to create multiple plots?**  
**Answer:**  
Subplots allow multiple plots in a single figure.  

### **Method 1: Using `plt.subplots()`**  
```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
axes[0, 0].plot(x1, y1, color="red")  # Top-left
axes[0, 1].scatter(x2, y2, color="blue")  # Top-right
axes[1, 0].bar(x3, y3, color="green")  # Bottom-left
axes[1, 1].hist(data, bins=20)  # Bottom-right
plt.tight_layout()  # Prevents overlapping
plt.show()
```

### **Method 2: Using `add_subplot()`**  
```python
fig = plt.figure(figsize=(10, 8))
ax1 = fig.add_subplot(2, 2, 1)  # (rows, cols, index)
ax1.plot(x1, y1)
ax2 = fig.add_subplot(2, 2, 2)
ax2.scatter(x2, y2)
plt.show()
```

**Key Parameters:**  
- `nrows`, `ncols`: Number of rows and columns.  
- `figsize`: Figure dimensions.  
- `tight_layout()`: Adjusts spacing.

## **11. Explain how to use Plotly’s `go.Figure()` for customizing plots.**  
**Answer:**  
`go.Figure()` provides fine-grained control over plots.  

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Bar(x=["A", "B", "C"], y=[10, 20, 30], name="Bar Chart"))
fig.add_trace(go.Scatter(x=["A", "B", "C"], y=[5, 15, 25], name="Line Plot"))

fig.update_layout(
    title="Customized Plot",
    xaxis_title="X-axis",
    yaxis_title="Y-axis",
    showlegend=True,
    template="plotly_dark"
)
fig.show()
```
**Key Features:**  
- Multiple traces (layers).  
- Custom layouts (titles, axes, themes).  
- Advanced interactivity.

## **12. Discuss the use of Plotly’s Dash for creating interactive web applications.**  
**Answer:**  
Dash is a framework for building analytical web apps using Plotly.  

### **Key Features:**  
- **Python-Based:** No JavaScript required.  
- **Interactive Components:** Sliders, dropdowns, buttons.  
- **Real-Time Updates:** Live data visualization.  
- **Deployment:** Can be hosted on servers.  

**Example:**  
```python
import dash
import dash_core_components as dcc
import dash_html_components as html
import plotly.express as px

app = dash.Dash()
app.layout = html.Div([
    dcc.Graph(figure=px.bar(df, x="x", y="y")),
    dcc.Dropdown(options=[{"label": "A", "value": "A"}])
])
app.run_server(debug=True)
```

## **17. When should you use a heatmap in your data visualizations, and how would you create one using Seaborn?**  
**Answer:**  
### **Heatmap Use Cases:**  
- **Correlation Matrix:** Relationships between variables.  
- **Categorical Data:** Frequency tables (e.g., confusion matrix).  
- **Time-Series Patterns:** Daily temperature trends.  

### **Creating a Heatmap:**  
```python
import seaborn as sns
flights = sns.load_dataset("flights")
flights_pivot = flights.pivot("month", "year", "passengers")

sns.heatmap(flights_pivot, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Passenger Count Heatmap")
plt.show()
```
**Parameters:**  
- `annot`: Displays values.  
- `fmt`: Format (e.g., `d` for integers).  
- `cmap`: Color map.

## **20. Explain when a stacked bar chart is useful and how you would create it in Matplotlib.**

- **Part-to-Whole Relationships:** Composition of categories (e.g., sales by product and region).  
- **Comparing Subgroups:** Breakdown within each category.

```python
labels = ['A', 'B', 'C']
men = [20, 35, 30]
women = [25, 32, 34]

plt.bar(labels, men, label="Men")
plt.bar(labels, women, bottom=men, label="Women")
plt.legend()
plt.show()
```
