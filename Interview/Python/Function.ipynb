# Python Interview Questions and Answers

## 1. What is the importance of functions in Python?

**Answer:**
Functions in Python are fundamental building blocks that provide several key benefits:

- **Code Reusability**: Functions allow you to write code once and reuse it multiple times, reducing duplication
- **Modularity**: They help break down complex problems into smaller, manageable parts
- **Abstraction**: Functions hide implementation details, allowing users to focus on functionality
- **Organization**: They make code more organized and readable
- **Scope Isolation**: Functions create their own namespace, preventing variable conflicts
- **Testing**: Individual functions can be tested separately
- **Collaboration**: Functions enable multiple developers to work on different parts simultaneously

Example:
```python
def calculate_area(radius):
    return 3.14 * radius ** 2

# Reuse the function multiple times
print(calculate_area(5))
print(calculate_area(7))
```

## 2. Explain the difference between default arguments and keyword arguments in Python.

**Answer:**

**Default Arguments**:
- Defined in the function definition with a default value
- Used when the caller doesn't provide a value for that parameter
- Must appear after non-default parameters in the function definition

**Keyword Arguments**:
- Passed to a function by explicitly naming each parameter and its value
- Allow passing arguments in any order
- Make function calls more readable

Example:
```python
# Function with default arguments
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

# Using default argument
greet("Alice")  # Uses default message "Hello"

# Using keyword arguments
greet(message="Hi", name="Bob")  # Order doesn't matter
```

Key Differences:
- Default arguments are defined in the function, keyword arguments are used when calling
- Default arguments provide fallback values, keyword arguments specify values explicitly
- Keyword arguments allow out-of-order parameter passing

## 3. What are *args and **kwargs in Python, and how are they used?

**Answer:**

`*args` and `**kwargs` are special syntaxes for handling variable numbers of arguments:

**`*args`**:
- Used to pass a variable number of non-keyword arguments
- Collects extra positional arguments as a tuple
- The name 'args' is convention, the asterisk (*) is what matters

**`**kwargs`**:
- Used to pass a variable number of keyword arguments
- Collects extra keyword arguments as a dictionary
- The name 'kwargs' is convention, the double asterisk (**) is what matters

Example:
```python
def example_function(arg1, *args, **kwargs):
    print(f"First argument: {arg1}")
    print(f"Additional positional arguments: {args}")
    print(f"Keyword arguments: {kwargs}")

example_function(1, 2, 3, 4, name="Alice", age=25)
```

Output:
```
First argument: 1
Additional positional arguments: (2, 3, 4)
Keyword arguments: {'name': 'Alice', 'age': 25}
```

Common Use Cases:
- When you need to create flexible functions that can handle different input patterns
- When writing wrapper or decorator functions
- When inheriting from classes where you don't know all method signatures in advance

## 4. How do iterators differ from generators? Provide examples.

**Answer:**

**Iterators**:
- Objects that implement the iterator protocol (__iter__() and __next__() methods)
- Return one element at a time when next() is called
- Maintain their state between calls
- Raise StopIteration when no more items are available
- All elements are generated at creation time (memory intensive for large sequences)

**Generators**:
- Special kind of iterator created using functions with yield statements
- Implement the iterator protocol automatically
- Generate values on-the-fly (lazy evaluation)
- Maintain state between yields
- More memory efficient for large sequences
- Can be created with generator expressions (similar to list comprehensions)

Example of Iterator:
```python
class Squares:
    def __init__(self, max):
        self.max = max
        self.n = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.n <= self.max:
            result = self.n ** 2
            self.n += 1
            return result
        else:
            raise StopIteration

squares = Squares(5)
for num in squares:
    print(num)
```

Example of Generator:
```python
def squares_gen(max):
    n = 0
    while n <= max:
        yield n ** 2
        n += 1

for num in squares_gen(5):
    print(num)

# Generator expression
squares = (x**2 for x in range(6))
for num in squares:
    print(num)
```

Key Differences:
1. Generators are simpler to implement (just use yield)
2. Generators automatically implement iterator protocol
3. Generators maintain state automatically
4. Generators are more memory efficient
5. Generators can only be iterated over once

## 5. Explain lambda functions in Python. How are they different from normal functions?

**Answer:**

**Lambda Functions** (anonymous functions):
- Small, single-expression functions defined with the `lambda` keyword
- Syntax: `lambda arguments: expression`
- Don't have a name (unless assigned to a variable)
- Can take any number of arguments but only one expression
- Return the result of the expression automatically
- Typically used for short, simple operations

Example:
```python
# Lambda function
square = lambda x: x * x
print(square(5))  # Output: 25

# Using lambda inline
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

**Normal Functions**:
- Defined with the `def` keyword
- Have a name
- Can contain multiple expressions and statements
- Can have docstrings
- Can include complex logic and control flow
- Return values explicitly with return statement

Example:
```python
def square(x):
    """Returns the square of x"""
    return x * x
```

Key Differences:
1. Syntax: lambda vs def
2. Lambda is limited to single expression
3. Lambda doesn't support statements (if, for, while, etc.)
4. Lambda functions don't have names (unless assigned)
5. Lambda functions can't have docstrings
6. Lambda functions are typically used for short, simple operations

## 6. How would you use map(), filter(), and reduce() in Python? Provide examples.

**Answer:**

**map(function, iterable)**:
- Applies a function to all items in an iterable
- Returns a map object (iterator) with the results

Example:
```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

**filter(function, iterable)**:
- Filters items from iterable based on function's return value (True/False)
- Returns a filter object (iterator) with items where function returns True

Example:
```python
numbers = [1, 2, 3, 4, 5, 6]
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)  # Output: [2, 4, 6]
```

**reduce(function, iterable)**:
- Applies a function cumulatively to items of iterable (from left to right)
- Reduces iterable to a single value
- Requires import from functools module

Example:
```python
from functools import reduce

numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 24 (1*2*3*4)
```

Practical Combined Example:
```python
from functools import reduce

data = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Filter even numbers, square them, then sum the squares
result = reduce(
    lambda x, y: x + y,
    map(
        lambda x: x**2,
        filter(lambda x: x % 2 == 0, data)
    )
)

print(result)  # Output: 120 (4 + 16 + 36 + 64)
```

**Answer:**

Using lambda expressions with `map()` and `filter()` provides several benefits:

   ```python
   # More concise than defining a separate function
   squared = list(map(lambda x: x**2, numbers))
   ```

2. **Readability**: When the operation is simple, lambda makes the code more readable by keeping the logic close to where it's used

3. **No Function Pollution**: Avoids creating many small named functions that might only be used once

4. **Functional Style**: Matches well with the functional programming style that map and filter encourage

5. **Immediate Context**: The operation's purpose is clear in the context where it's used

6. **No Namespace Clutter**: Doesn't add names to the namespace that might conflict or never be used again

Example without lambda would require separate functions:
```python
def square(x):
    return x**2

def is_even(x):
    return x % 2 == 0

squared = list(map(square, numbers))
evens = list(filter(is_even, numbers))
```

However, lambdas should be used judiciously:
- For complex operations, regular functions are better
- If the same operation is used multiple times, a named function is better
- Lambda expressions shouldn't be so complex that they hurt readability

## 8. What are classes and objects in Python? Explain with examples.

**Answer:**

**Classes**:
- Blueprints or templates for creating objects
- Define attributes (data) and methods (functions) that the objects will have
- Created using the `class` keyword
- Follow PascalCase naming convention by convention

**Objects**:
- Instances of classes
- Concrete entities created from class blueprints
- Each object has its own set of attributes and can access class methods
- Created by calling the class like a function

Example:
```python
class Dog:
    # Class attribute (shared by all instances)
    species = "Canis familiaris"

    # Initializer (constructor)
    def __init__(self, name, age):
        # Instance attributes
        self.name = name
        self.age = age
    
    # Instance method
    def bark(self):
        print(f"{self.name} says woof!")

    def describe(self):
        print(f"{self.name} is {self.age} years old.")

# Creating objects (instances)
dog1 = Dog("Buddy", 5)
dog2 = Dog("Milo", 3)

# Accessing attributes and methods
print(dog1.name)  # Output: Buddy
print(dog2.age)   # Output: 3
dog1.bark()       # Output: Buddy says woof!
dog2.describe()   # Output: Milo is 3 years old.

# Class attribute is shared
print(dog1.species)  # Output: Canis familiaris
print(dog2.species)  # Output: Canis familiaris
```

Key Concepts:
- `__init__()`: Special method called when object is created (constructor)
- `self`: Reference to the current instance (first parameter of instance methods)
- Instance attributes: Unique to each object (name, age in example)
- Class attributes: Shared by all instances (species in example)
- Methods: Functions defined within a class that operate on instances

## 9. Explain the concept of inheritance in Python. How does it work?

**Answer:**

Inheritance is a fundamental OOP concept where a class (child/derived class) inherits attributes and methods from another class (parent/base class). This promotes code reuse and establishes hierarchical relationships.

Key Aspects:
1. **Base Class (Parent)**: The class being inherited from
2. **Derived Class (Child)**: The class that inherits
3. Child class can:
   - Inherit all parent's attributes and methods
   - Override parent methods with its own implementations
   - Add new attributes and methods
   - Extend parent methods using `super()`

Example:
```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def speak(self):
        print(f"{self.name} barks")

class Cat(Animal):
    def speak(self):
        print(f"{self.name} meows")

    def purr(self):
        print(f"{self.name} is purring")

# Creating instances
animal = Animal("Generic Animal")
dog = Dog("Buddy")
cat = Cat("Whiskers")

animal.speak()  # Output: Generic Animal makes a sound
dog.speak()     # Output: Buddy barks
cat.speak()     # Output: Whiskers meows
cat.purr()      # Output: Whiskers is purring
```

How Inheritance Works:
1. Method Resolution Order (MRO) determines how Python searches for attributes and methods
2. When an attribute/method is accessed:
   - Python first checks the instance
   - Then the child class
   - Then parent classes (in MRO order)
3. `super()` allows accessing parent class methods
4. Multiple inheritance is supported (though should be used carefully)

Benefits:
- Code reuse
- Hierarchical organization
- Polymorphism (objects of different classes can be treated uniformly)
- Extensibility (easy to add specialized classes)

## 10. What is the difference between single inheritance and multiple inheritance in Python?

**Answer:**

**Single Inheritance**:
- A class inherits from only one parent class
- Forms a simple, linear hierarchy
- Less complex and easier to understand
- No ambiguity in method resolution

Example:
```python
class Parent:
    pass

class Child(Parent):  # Single inheritance
    pass
```

**Multiple Inheritance**:
- A class inherits from more than one parent class
- Forms a complex, diamond-shaped hierarchy
- More powerful but can lead to complexity
- Method resolution follows the MRO (Method Resolution Order) algorithm
- Potential for the "diamond problem" (ambiguity when same method exists in multiple parents)

Example:
```python
class Father:
    def skills(self):
        print("Gardening, Programming")

class Mother:
    def skills(self):
        print("Cooking, Art")

class Child(Father, Mother):  # Multiple inheritance
    def skills(self):
        Father.skills(self)
        Mother.skills(self)
        print("Sports")

child = Child()
child.skills()
```

Key Differences:

| Aspect              | Single Inheritance            | Multiple Inheritance              |
|---------------------|-------------------------------|------------------------------------|
| Parent Classes      | One                           | Multiple                          |
| Complexity          | Simple                        | Complex                           |
| Method Resolution   | Straightforward              | Uses MRO algorithm                |
| Ambiguity           | None                          | Possible (diamond problem)        |
| Use Cases           | Most common, simpler systems  | Mixins, complex interfaces        |
| Performance         | Slightly faster              | Slightly slower due to MRO lookup |

Python handles multiple inheritance with:
1. Method Resolution Order (MRO) - determines the order in which base classes are searched
2. `super()` function - cooperatively calls methods in the inheritance chain
3. C3 Linearization algorithm - ensures consistent method resolution order

Best Practices for Multiple Inheritance:
- Use it judiciously (prefer composition when possible)
- Keep inheritance hierarchies shallow
- Use mixins for adding specific behaviors
- Document the expected method resolution clearly

## 11. What is encapsulation, and how is it implemented in Python?

**Answer:**

Encapsulation is an OOP concept that bundles data (attributes) and methods that operate on that data within a single unit (class), while restricting direct access to some components.

**Implementation in Python**:

1. **Public Members**:
   - Default in Python
   - Accessible from anywhere
   - No special syntax

2. **Protected Members**:
   - Conventionally indicated by single underscore prefix: `_name`
   - Not enforced by Python interpreter (just a convention)
   - Signals "internal use only" to other developers

3. **Private Members**:
   - Indicated by double underscore prefix: `__name`
   - Name mangling makes it harder (but not impossible) to access outside class
   - Actually becomes `_ClassName__name`

Example:
```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder  # Public
        self._account_number = "123456789"   # Protected
        self.__balance = balance            # Private
    
    # Public method to access private attribute
    def get_balance(self):
        return self.__balance
    
    # Public method to modify private attribute safely
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False

account = BankAccount("Alice", 1000)

# Accessing public member
print(account.account_holder)  # Works

# Accessing protected member (convention only)
print(account._account_number)  # Works but shouldn't

# Accessing private member directly fails
# print(account.__balance)  # AttributeError

# Accessing private member via name mangling
print(account._BankAccount__balance)  # Works but shouldn't be done

# Proper way to access via public method
print(account.get_balance())  # Correct approach
```

Key Aspects of Encapsulation in Python:
1. **Data Hiding**: Making attributes private/protected to prevent direct modification
2. **Access Control**: Providing controlled access via methods (getters/setters)
3. **Validation**: Methods can validate data before modifying attributes
4. **Flexibility**: Implementation can change without affecting external code

Python's Philosophy:
- "We're all consenting adults here" - trusts developers to respect conventions
- No strict enforcement of access control
- Provides tools (underscores, name mangling) to indicate intended usage

Best Practices:
- Use private members for implementation details that should never be accessed directly
- Use protected members for implementation details that might need to be accessed by subclasses
- Provide public methods for safe interaction with the object's state
- Use property decorators for elegant attribute access control

## 12. What are static methods and class methods? Explain with an example.

**Answer:**

**Static Methods**:
- Defined using `@staticmethod` decorator
- Don't receive implicit first argument (no `self` or `cls`)
- Can't access or modify class or instance state
- Essentially regular functions but belong to class's namespace
- Called using class name or instance

**Class Methods**:
- Defined using `@classmethod` decorator
- Receive class as implicit first argument (conventionally named `cls`)
- Can access and modify class state but not instance state
- Often used as factory methods for creating instances
- Called using class name or instance

Example:
```python
class Pizza:
    def __init__(self, ingredients):
        self.ingredients = ingredients
    
    def __repr__(self):
        return f"Pizza({self.ingredients!r})"
    
    @classmethod
    def margherita(cls):
        return cls(["mozzarella", "tomatoes"])
    
    @classmethod
    def prosciutto(cls):
        return cls(["mozzarella", "tomatoes", "ham"])
    
    @staticmethod
    def area(radius):
        return 3.14 * radius ** 2

# Using class methods as factory methods
margherita = Pizza.margherita()
print(margherita)  # Output: Pizza(['mozzarella', 'tomatoes'])

prosciutto = Pizza.prosciutto()
print(prosciutto)  # Output: Pizza(['mozzarella', 'tomatoes', 'ham'])

# Using static method
print(Pizza.area(12))  # Output: 452.16

# Can also call from instance
p = Pizza(["cheese"])
print(p.area(10))  # Output: 314.0
```

Key Differences:

| Aspect          | Instance Method          | Class Method               | Static Method             |
|----------------|-------------------------|---------------------------|--------------------------|
| Decorator      | None                    | @classmethod              | @staticmethod            |
| First Param    | self (instance)         | cls (class)               | None                     |
| Access to      | Instance and class      | Class only                | Neither                  |
| Can modify     | Instance and class state| Class state only          | Nothing                  |
| Typical Use    | Behavior tied to instance| Factory methods, alternate constructors | Utility functions related to class |

When to Use:
- **Class Methods**:
  - When you need an alternate constructor (factory pattern)
  - When the method needs to know about the class but not instances
  - For methods that need to work with the class itself

- **Static Methods**:
  - For utility functions that are logically related to the class
  - When the method doesn't need access to instance or class state
  - To group related functions within a class namespace

## 13. What are dunder methods in Python? Why are they useful?

**Answer:**

Dunder methods (double underscore methods) are special methods in Python that begin and end with double underscores (e.g., `__init__`, `__str__`). They allow classes to implement and interact with Python's built-in features.

**Key Dunder Methods and Their Uses**:

1. **Object Initialization**:
   - `__init__`: Constructor, called when object is created
   - `__new__`: Actually creates the instance (before `__init__`)

2. **String Representation**:
   - `__str__`: Informal string representation (used by `str()` and `print()`)
   - `__repr__`: Official string representation (used by `repr()`)

3. **Comparison Operations**:
   - `__eq__`, `__lt__`, `__gt__`, etc.: Enable comparison operators

4. **Arithmetic Operations**:
   - `__add__`, `__sub__`, `__mul__`, etc.: Enable arithmetic operators

5. **Container Methods**:
   - `__len__`: Implements `len()`
   - `__getitem__`, `__setitem__`: Enable indexing
   - `__contains__`: Implements `in` operator

6. **Callable Objects**:
   - `__call__`: Makes instances callable like functions

7. **Context Managers**:
   - `__enter__`, `__exit__`: Enable `with` statement

Example:
```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    def __len__(self):
        return 2
    
    def __abs__(self):
        return (self.x**2 + self.y**2)**0.5

v1 = Vector(2, 4)
v2 = Vector(1, 3)

print(v1 + v2)  # Vector(3, 7) - uses __add__
print(v1 == v2) # False - uses __eq__
print(len(v1))  # 2 - uses __len__
print(abs(v1))  # 4.472 - uses __abs__
```

Why Dunder Methods Are Useful:
1. **Operator Overloading**: Allow objects to work with Python operators
2. **Integration with Python Ecosystem**: Enable objects to work seamlessly with built-in functions
3. **Polymorphism**: Different classes can implement common interfaces
4. **Custom Behavior**: Define how objects should behave in various contexts
5. **Emulation of Built-in Types**: Make user-defined objects work like Python built-ins
6. **Context Management**: Enable resource management with `with` statements

Key Points:
- Dunder methods are how Python implements polymorphism
- They're automatically called by Python in specific situations
- They make user-defined objects behave like built-in types
- They're the foundation of Python's data model

## 14. What is the difference between method overriding and method overloading?

**Answer:**

**Method Overriding**:
- Occurs in inheritance when a child class provides a specific implementation of a method already defined in its parent class
- The method in child class has the same name and parameters as in parent
- Runtime polymorphism - which version to call is determined at runtime based on object type
- Supported in Python

Example:
```python
class Parent:
    def show(self):
        print("Parent's show")

class Child(Parent):
    def show(self):  # Method overriding
        print("Child's show")

p = Parent()
c = Child()

p.show()  # Output: Parent's show
c.show()  # Output: Child's show
```

**Method Overloading**:
- Having multiple methods with the same name but different parameters in the same class
- Compile-time polymorphism - which version to call is determined at compile time
- Not directly supported in Python (no compile-time type checking)
- Can be simulated using default arguments or variable arguments

Example (simulated in Python):
```python
class Example:
    def add(self, a, b, c=None):
        if c is not None:
            return a + b + c
        return a + b

e = Example()
print(e.add(2, 3))      # Output: 5
print(e.add(2, 3, 4))   # Output: 9
```

Key Differences:

| Aspect              | Method Overriding                  | Method Overloading                |
|---------------------|------------------------------------|-----------------------------------|
| Definition          | Same method in child class         | Multiple methods in same class    |
| Parameters          | Same signature                     | Different parameters              |
| Polymorphism Type   | Runtime                            | Compile-time                      |
| Inheritance         | Requires inheritance               | Within same class                 |
| Python Support      | Fully supported                    | Not directly supported            |
| Resolution          | Based on object type at runtime    | Based on arguments at compile time|

Python Approach:
- Python doesn't support traditional method overloading like Java/C++
- Instead, use:
  - Default parameters
  - Variable arguments (*args, **kwargs)
  - Type checking and different behavior based on input types
  - Single method that handles different cases

Example of Pythonic "overloading":
```python
class MathOperations:
    def add(self, *args):
        if len(args) == 2:
            return args[0] + args[1]
        elif len(args) == 3:
            return args[0] + args[1] + args[2]
        else:
            return sum(args)

m = MathOperations()
print(m.add(2, 3))      # 5
print(m.add(2, 3, 4))   # 9
print(m.add(1, 2, 3, 4)) # 10
```

## 15. Explain the use of decorators in Python and provide an example.

**Answer:**

Decorators are a powerful Python feature that allows modifying or extending the behavior of functions or methods without permanently changing them. They provide a clean, elegant syntax for applying transformations.

**Key Characteristics**:
- Functions that take another function as input and return a modified function
- Use the `@decorator_name` syntax above function definitions
- Can be chained (multiple decorators on one function)
- Preserve the original function's metadata (unless fixed with `functools.wraps`)
- Commonly used for:
  - Logging
  - Timing
  - Access control/authentication
  - Caching/memoization
  - Input validation
  - Retry mechanisms

Basic Example:
```python
def simple_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@simple_decorator
def say_hello():
    print("Hello!")

say_hello()
```

Output:
```
Before function execution
Hello!
After function execution
```

Practical Example - Timing Decorator:
```python
import time
from functools import wraps

def timer(func):
    @wraps(func)  # Preserves function metadata
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"{func.__name__} executed in {end_time - start_time:.4f} seconds")
        return result
    return wrapper

@timer
def expensive_operation(n):
    """Simulates an expensive operation"""
    time.sleep(n)
    return "Done"

result = expensive_operation(2)
print(result)
```

Output:
```
expensive_operation executed in 2.0023 seconds
Done
```

Class-Based Decorator Example:
```python
class CountCalls:
    def __init__(self, func):
        self.func = func
        self.calls = 0
    
    def __call__(self, *args, **kwargs):
        self.calls += 1
        print(f"Call {self.calls} of {self.func.__name__}")
        return self.func(*args, **kwargs)

@CountCalls
def greet(name):
    print(f"Hello, {name}")

greet("Alice")
greet("Bob")
```

Output:
```
Call 1 of greet
Hello, Alice
Call 2 of greet
Hello, Bob
```

Decorator Use Cases:
1. **Logging**: Track function calls and arguments
2. **Authentication**: Verify permissions before execution
3. **Caching**: Store results of expensive computations
4. **Validation**: Check input/output formats
5. **Retry Logic**: Automatically retry failed operations
6. **Rate Limiting**: Control how often a function can be called

Key Points:
- Decorators are higher-order functions
- They follow the decorator pattern from software design
- They provide a way to modify behavior without changing original code
- Multiple decorators are applied from bottom to top (nearest to function first)
