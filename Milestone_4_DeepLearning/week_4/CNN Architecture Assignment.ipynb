## 1. What is the role of filters and feature maps in Convolutional Neural Network (CNN)?

**Filters** (also called **kernels**) and **feature maps** are the heart of a **Convolutional Neural Network (CNN)**.

Here’s their role:

1. **Filters:**

   * A filter is a small matrix (like 3×3 or 5×5) that slides over the image.
   * It helps the network focus on specific patterns such as **edges, corners, textures**, etc.
   * Each filter learns to detect a **different type of feature**.

2. **Feature Maps:**

   * After a filter slides over the image and does convolution, the result is a **feature map** (or activation map).
   * The feature map shows **where in the image** that particular pattern (detected by the filter) is found.

**In short:**

* Filters = pattern detectors
* Feature maps = show the presence and location of those patterns in the image.

## 2. Explain the concepts of padding and stride in CNNs(Convolutional Neural Network). How do they affect the output dimensions of feature maps?

### **1. Padding**

* **Meaning:** Padding means adding extra rows and columns (usually filled with zeros) around the input image.
* **Why it’s used:**

  * To keep the **output size same** as the input size.
  * To make sure filters can scan **edges and corners** of the image.
* **Types of padding:**

  * **Valid padding:** No padding (output becomes smaller).
  * **Same padding:** Adds zeros so output size stays the same as input.

**Effect on output:**
More padding → larger output feature map.
No padding → smaller output feature map.

---

### **2. Stride**

* **Meaning:** Stride is how many pixels the filter moves each time when sliding over the image.
* **Example:**

  * **Stride = 1:** filter moves one pixel at a time → larger output.
  * **Stride = 2:** filter jumps two pixels → smaller output.

**Effect on output:**
Higher stride → smaller output feature map (because fewer positions are checked).
Lower stride → larger output feature map.


## 3. Define receptive field in the context of CNNs. Why is it important for deep architectures?


### **Receptive Field – Definition**

In a **Convolutional Neural Network (CNN)**, the **receptive field** is the **region of the input image** that a particular neuron (or node) in a layer “looks at” or is **influenced by**.

In other words:
The receptive field tells **how much of the original image** affects a neuron’s output.

---

### **Example:**

* In the **first layer**, each neuron might see only a small 3×3 area (the filter size).
* In **deeper layers**, because each layer combines information from the previous one, neurons start seeing a **larger portion** of the image — this is a **larger receptive field**.

---

### **Why it’s important for deep architectures**

1. **Captures global context:**
   Deeper layers with larger receptive fields can understand **bigger and more complex patterns**, like shapes or objects, not just edges.

2. **Improves feature understanding:**
   Early layers → detect small features (edges, textures)
   Later layers → detect higher-level features (faces, objects, etc.)

3. **Better performance:**
   A properly designed receptive field helps the model recognize features regardless of where they appear in the image.

---

**In short:**

* **Receptive field:** area of input that affects one neuron.
* **Importance:** allows deeper CNNs to understand **larger and more meaningful patterns** in images.


## 4.  Discuss how filter size and stride influence the number of parameters in a CNN.

**Filter Size and Stride in CNNs**

**Filter Size:**
The filter (kernel) size determines how many weights a filter has. Each filter’s parameters depend on its height, width, and the number of input channels.
A larger filter (like 5×5) has more parameters than a smaller one (like 3×3). More parameters mean higher computation and a greater risk of overfitting. This is why modern CNNs often use small filters, such as 3×3, stacked in multiple layers.

**Stride:**
Stride controls how far the filter moves each time it slides over the image. It does not affect the number of parameters, but it does change the size of the output feature map.
A larger stride produces a smaller output (fewer activations), while a smaller stride gives a larger output with more detail.

**In summary:**

* Filter size affects the number of parameters.
* Stride affects the output size, not the parameters.

## 5. Compare and contrast different CNN-based architectures like LeNet, AlexNet, and VGG in terms of depth, filter sizes, and performance.

**Comparison of LeNet, AlexNet, and VGG Architectures**

**1. LeNet (1998):**

* **Depth:** Very shallow, around 5–7 layers.
* **Filter Size:** Mostly large filters (5×5).
* **Performance:** Designed for handwritten digit recognition (MNIST). Simple and efficient for small grayscale images but not suitable for large, complex datasets.

**2. AlexNet (2012):**

* **Depth:** Deeper than LeNet, about 8 layers (5 convolutional + 3 fully connected).
* **Filter Size:** Starts with large filters (11×11), then smaller ones (5×5, 3×3).
* **Performance:** First CNN to achieve breakthrough results on ImageNet. Introduced ReLU activation, dropout, and GPU training for faster computation.

**3. VGG (2014):**

* **Depth:** Much deeper, with 16 or 19 layers.
* **Filter Size:** Uses only small filters (3×3) throughout the network.
* **Performance:** Achieves high accuracy and strong feature extraction but is computationally heavy and memory-intensive.

**In summary:**

| Model       | Depth                | Filter Size         | Key Traits                                   |
| ----------- | -------------------- | ------------------- | -------------------------------------------- |
| **LeNet**   | Shallow (5–7 layers) | Large (5×5)         | Simple, early CNN, good for small images     |
| **AlexNet** | Moderate (8 layers)  | Mixed (11×11 → 3×3) | Introduced ReLU and GPU training             |
| **VGG**     | Deep (16–19 layers)  | Small (3×3)         | Very accurate, but computationally expensive |


```python
# 6. Using keras, build and train a simple CNN model on the MNIST dataset from scratch. 
# Include code for module creation, compilation, training, and evaluation.

# Import necessary libraries
import tensorflow as tf
from tensorflow.keras import datasets, layers, models

# 1. Load and preprocess the MNIST dataset
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

# Reshape to include channel dimension (28x28x1 for grayscale)
x_train = x_train.reshape((x_train.shape[0], 28, 28, 1))
x_test = x_test.reshape((x_test.shape[0], 28, 28, 1))

# Normalize pixel values to [0, 1]
x_train, x_test = x_train / 255.0, x_test / 255.0

# 2. Build the CNN model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')  # 10 classes for digits 0–9
])

# 3. Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 4. Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1)

# 5. Evaluate the model
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {test_acc:.4f}")
```

```python
# 7. Load and preprocess the CIFAR-10 dataset using Keras, and create a CNN model to classify RGB images. 
# Show your preprocessing and architecture.

import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt

# Load dataset
(x_train, y_train), (x_test, y_test) = datasets.cifar10.load_data()

# Normalize pixel values (0–255 → 0–1)
x_train, x_test = x_train / 255.0, x_test / 255.0

# Convert labels to one-hot vectors (optional for categorical crossentropy)
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)

print("Training data shape:", x_train.shape)
print("Test data shape:", x_test.shape)

model = models.Sequential([
    # Convolution + Pooling Block 1
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    layers.MaxPooling2D((2, 2)),

    # Convolution + Pooling Block 2
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Convolution + Pooling Block 3
    layers.Conv2D(64, (3, 3), activation='relu'),

    # Flatten and Fully Connected Layers
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train, epochs=10,
                    validation_data=(x_test, y_test),
                    batch_size=64)

test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"Test Accuracy: {test_acc:.4f}")
```

```python
# 8. Using PyTorch, write a script to define and train a CNN on the MNIST dataset. 
# Include model definition, data loaders, training loop, and accuracy evaluation.

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Transform: convert to tensor and normalize
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Download training and test datasets
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

# Data loaders
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

class CNNModel(nn.Module):
    def __init__(self):
        super(CNNModel, self).__init__()
        self.conv1 = nn.Conv2d(1, 32, 3, 1)    # 1 input channel (grayscale), 32 filters
        self.conv2 = nn.Conv2d(32, 64, 3, 1)
        self.fc1 = nn.Linear(9216, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = F.relu(self.conv1(x))
        x = F.max_pool2d(F.relu(self.conv2(x)), 2)
        x = torch.flatten(x, 1)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return F.log_softmax(x, dim=1)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = CNNModel().to(device)

optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.CrossEntropyLoss()

epochs = 5

for epoch in range(epochs):
    model.train()
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)

        optimizer.zero_grad()
        output = model(data)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}")

model.eval()
correct = 0
total = 0

with torch.no_grad():
    for data, target in test_loader:
        data, target = data.to(device), target.to(device)
        output = model(data)
        _, predicted = torch.max(output.data, 1)
        total += target.size(0)
        correct += (predicted == target).sum().item()

accuracy = 100 * correct / total
print(f"Test Accuracy: {accuracy:.2f}%")
```
