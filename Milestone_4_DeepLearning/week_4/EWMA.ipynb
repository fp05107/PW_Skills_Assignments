```python
import pandas as pd

# Sample data
data = {
    'Day': range(1, 11),
    'Price': [50, 52, 53, 54, 52, 51, 50, 49, 48, 47]
}

df = pd.DataFrame(data)

# Calculate EWMA manually (span = 3 → alpha = 0.5)
alpha = 2 / (3 + 1)  # 0.5
ema = [df.loc[0, 'Price']]  # start with first price as initial EMA

for price in df['Price'][1:]:
    next_ema = alpha * price + (1 - alpha) * ema[-1]
    ema.append(next_ema)

# Add EWMA to DataFrame
df['EWMA'] = ema

# Loop through each row and print EWMA value
for index, row in df.iterrows():
    print(f"Day {row['Day']} → EWMA: {row['EWMA']:.2f}")
```
