```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
```

```python
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()
```

```python
# Preprocess the data
# LeNet expects input shape (32, 32, 1) but MNIST is (28, 28, 1)
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
```

```python
x_train.shape
```

```python
x_train = tf.pad(x_train, [[0,0], [2,2], [2,2], [0,0]], "CONSTANT")
x_test = tf.pad(x_test, [[0,0],[2,2],[2,2],[0,0]], "CONSTANT")
```

```python
# Normalize pixel values for faster training
x_train, x_test = x_train / 255, x_test / 255
```

```python
# LeNet-5 Consists of:
# Conv -> Pool -> Conv -> Pool -> Fully Connected layers -> Output
model = models.Sequential([
    # Layer 1: Convolution (6 filters of 5x5)
    # Input shape = (32x32x1)
    layers.Conv2D(6, kernel_size=(5,5), activation='tanh', input_shape=(32,32,1)),

    # Layer 2: Average Pooling (reduces spatial size)
    layers.AveragePooling2D(pool_size=3),

    # Layer 3: Convolution (16 filters of 5x5)
    layers.Conv2D(16, kernel_size=(5,5), activation='tanh'),

    # Layer 4: Another Average Pooling layer
    layers.AveragePooling2D(pool_size=3),

    # Flatten the feature maps into a 1D vector
    layers.Flatten(),

    # Layer 5: Fully connected layer with 120 neurons
    layers.Dense(120, activation='tanh'),

    # Layer 6: Fully connected layer with 84 neurons
    layers.Dense(84, activation='tanh'),

    # Output layer: 10 neurons (digits 0–9), softmax gives probabilities
    layers.Dense(10, activation='softmax')
])
```

```python
# Define optimizer, loss function, and metrics for training
model.compile(
    optimizer='adam',                        # Adam optimizer for adaptive learning
    loss='sparse_categorical_crossentropy',  # Suitable for integer labels
    metrics=['accuracy']                     # Track accuracy
)
```

```python
# Train for 10 epochs, use 10% of training data for validation
model.fit(
    x_train, y_train,
    epochs=1,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)
```

```python
# -----------------------------------------------------------
# 6️⃣ Evaluate the model on test data
# -----------------------------------------------------------
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"\n✅ Test accuracy: {test_acc:.4f}")
```
