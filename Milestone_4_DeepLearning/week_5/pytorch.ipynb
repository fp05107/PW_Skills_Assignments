```python
import torch
import time

# Size of the matrices
N = 10000

# CPU tensors
a_cpu = torch.rand((N, N))
b_cpu = torch.rand((N, N))

# GPU tensors (if available)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
a_gpu = a_cpu.to(device)
b_gpu = b_cpu.to(device)

# CPU matmul timing
start = time.time()
c_cpu = torch.matmul(a_cpu, b_cpu)
cpu_time = time.time() - start
print(f"CPU matmul time: {cpu_time:.4f} seconds")

# GPU matmul timing
# Make sure to synchronize before and after timing
torch.cuda.synchronize()  # Wait for GPU
start = time.time()
c_gpu = torch.matmul(a_gpu, b_gpu)
torch.cuda.synchronize()  # Wait for GPU to finish
gpu_time = time.time() - start
print(f"GPU matmul time: {gpu_time:.4f} seconds")

# Optional: Verify results are close
print("Are results close?", torch.allclose(c_cpu, c_gpu.cpu(), atol=1e-6))
```
