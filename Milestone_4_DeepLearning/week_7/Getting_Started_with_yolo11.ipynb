```python
import cv2
cv2_imshow = cv2.imshow

def show_image(image_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Error: Could not read image from {image_path}")
        return
    cv2_imshow("Image", image)
    cv2.waitKey(0)
```
