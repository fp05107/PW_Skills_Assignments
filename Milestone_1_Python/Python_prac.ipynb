```python
# Write a program that will determine weather when the value of temperature and humidity is provided by the user.
# TEMPERATURE(C)      HUMIDITY(%)      WEATHER

#       >= 30                             >=90                Hot and Humid
#       >= 30                             < 90                 Hot
#       <30                                >= 90               Cool and Humid
#       <30                                 <90                 Cool

temp = int(input("Enter Temperature: "))
humidity = int(input("Enter Humidity: "))

match (temp >= 30, humidity >= 90):
    case (True, True):
        print("Hot and Humid")
    case (True, False):
        print("Hot")
    case (False, True):
        print("Cool and Humid")
    case (False, False):
        print("Cool")
```

```python
# Write a program that will check whether the number is armstrong number or not.


def convert_to_armstrong_num(num):
    list = []
    str_num = str(num)
    # print(str_num)
    arm_strong_num = 0
    for i in range(0, len(str_num)):
        list.append(str_num[i])

    # print(list)
    for item in list:
        arm_strong_num += int(item)**len(list)
    
    if(num == arm_strong_num):
        print("Yes....")
    else:
        print("No........")
    print(arm_strong_num)

convert_to_armstrong_num(370)
```

```python
# Write a program that will give you the in hand salary after deduction of HRA(10%),DA(5%),PF(3%), 
# and tax(if salary is between 5-10 lakh–10%),(11-20lakh–20%),(20< _   – 30%)(0-1lakh print k).


def calculate_inhand_salary(salary):
    inhand_salary = 0
    hra = salary * 10/100
    da = salary * 5/100
    pf = salary * 3/100
    tax = 0 if salary < 500000 else 0.1*salary if salary <= 1000000 else 0.2*salary if salary <= 2000000 else 0.3*salary

    inhand_salary = salary - (hra + da + pf + tax)
    print(inhand_salary)

calculate_inhand_salary(500000)
```
