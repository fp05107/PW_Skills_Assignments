# Theoretical Questions

## 1. What is unsupervised learning in the context of machine learning?

Unsupervised learning is a type of machine learning where the model is trained on data that has no labels. The goal is to identify patterns, structures, or groupings within the data without prior knowledge of the outcomes. Common techniques include clustering (like K-Means) and dimensionality reduction (like PCA).


## 2. How does K-Means clustering algorithm work?

K-Means Clustering works in the following steps:
- Choose K - Decide the number of cluster you want.
- Initialize centroids - Randomly pick K points as the initial cluster centers.
- Assign Points - Each data point is assigned to the nearest centroid (Based on Distance)
- Update Centroids - For each cluster calculate the mean of all points in that cluster and move the centroid to this mean
- Repeat - Reassign points and update centroids until centroids no longer change significantly or a maximum number of iterations is reached.

The result is K clusters where points in the same cluster are more similar to each other than to points in other clusters.

## 3. Explain the concept of a dendrogram in hierarchical clustering.

A **dendrogram** is a tree-like diagram that shows how data points are grouped together step by step in **hierarchical clustering**.

* Each **leaf node** at the bottom represents a single data point.
* As you move up, branches join together to form clusters.
* The **height** at which two branches merge indicates the distance or dissimilarity between the clusters being joined.
* By “cutting” the dendrogram at a chosen height, you can decide the number of clusters to form.

In short, a dendrogram visually represents the process of hierarchical clustering and helps decide the optimal number of clusters.


## 4. What is the main difference between K-Means and Hierarchical Clustering?

The main difference is in how they form clusters:

* **K-Means**: A **partitioning method**. You must predefine the number of clusters (K). It assigns points to clusters and adjusts centroids iteratively until convergence.

* **Hierarchical Clustering**: A **hierarchy-building method**. It does not require specifying the number of clusters upfront. Instead, it creates a dendrogram that shows nested groupings, and you can choose the number of clusters by cutting the tree at a certain level.

In short: K-Means is fast and requires K as input, while Hierarchical clustering builds a tree of clusters without needing K initially.

## 5. What are the advantages of DBSCAN over K-Means?

1. **No need to specify number of clusters (K):**

   * K-Means requires you to set K beforehand.
   * DBSCAN automatically finds the number of clusters based on data density.

2. **Can find arbitrarily shaped clusters:**

   * K-Means assumes clusters are spherical around centroids.
   * DBSCAN can detect clusters of any shape (e.g., elongated, curved).

3. **Handles noise/outliers well:**

   * K-Means forces every point into a cluster, even outliers.
   * DBSCAN labels low-density points as **noise** instead of forcing them.

4. **More robust with varying cluster sizes and densities (to some extent):**

   * K-Means struggles when clusters are not uniform in size/density.
   * DBSCAN works better in such cases.

In short: **DBSCAN is better when you don’t know the number of clusters, expect non-spherical clusters, or need to handle noise.**

## 6.  When would you use Silhouette Score in clustering?

You would use the **Silhouette Score** in clustering to **evaluate how well the data points are grouped into clusters**.

It measures two things for each point:

1. **Cohesion** – how close the point is to others in its own cluster.
2. **Separation** – how far the point is from points in other clusters.

The score ranges from **-1 to +1**:

* Close to **+1** → point is well matched to its cluster and far from others.
* Around **0** → point lies between clusters (uncertain assignment).
* Close to **-1** → point may be assigned to the wrong cluster.

In practice, you use the **average Silhouette Score** of all points to:

* Compare clustering quality across different algorithms (e.g., K-Means vs DBSCAN).
* Decide the **optimal number of clusters (K)** by checking which K gives the best score.

## 7.  What are the limitations of Hierarchical Clustering?


1. **Scalability Issues**

   * Time complexity is $O(n^2)$ or worse, making it inefficient for very large datasets.

2. **No Reassignment Once Merged/Split**

   * Once two clusters are merged (or split), the decision cannot be undone.
   * This can lead to suboptimal clustering if an early mistake happens.

3. **Sensitive to Noise and Outliers**

   * Outliers can distort the cluster structure, since all points must be included.

4. **Choice of Linkage Method Matters**

   * Different linkage methods (single, complete, average, ward) can produce very different dendrograms.

5. **Difficulty Handling High-Dimensional Data**

   * In high dimensions, distance metrics become less meaningful (curse of dimensionality).

6. **No Clear Objective Function**

   * Unlike K-Means, which minimizes within-cluster variance, hierarchical clustering doesn’t optimize a specific global criterion.

In short: hierarchical clustering is easy to interpret and works well for small datasets, but it’s **slow, sensitive to noise, and less flexible** compared to methods like K-Means or DBSCAN.

## 8. Why is feature scaling important in clustering algorithms like K-Means?

Feature scaling is important in clustering algorithms like **K-Means** because the algorithm relies on **distance calculations (usually Euclidean distance)** to assign points to clusters.

If features are on very different scales:

* A feature with large numerical values (e.g., income in dollars) will **dominate** the distance measure.
* A feature with smaller values (e.g., age in years) will have little effect.
* This leads to biased clustering results.

By scaling (e.g., **standardization** or **min-max normalization**):

* All features contribute **equally** to the distance metric.
* The algorithm forms clusters based on the true structure of the data, not just the scale of features.

**Example:** If one feature ranges from 1–1000 and another from 1–10, without scaling the first feature will overshadow the second, even if both are equally important.

In short: **scaling ensures fair contribution of all features in clustering.**

## 9. How does DBSCAN identify noise points?

DBSCAN identifies **noise points** (also called outliers) based on two parameters:

* **ε (epsilon):** the neighborhood radius around a point.
* **MinPts:** the minimum number of points required within ε to form a dense region (a cluster).

### Steps:

1. A point is classified as:

   * **Core Point**: if it has at least MinPts points (including itself) within ε.
   * **Border Point**: if it has fewer than MinPts within ε but lies within the neighborhood of a core point.
   * **Noise Point**: if it is neither a core point nor a border point.

2. **Noise points** are those that do not belong to any cluster because they don’t satisfy the density requirement and aren’t within reach of a core point.

### Example:

If ε = 0.5 and MinPts = 5 → any point with fewer than 5 neighbors in a 0.5 radius, and not connected to a core point, will be marked as **noise**.

In short: **DBSCAN labels points as noise if they are in low-density regions, isolated from clusters.**


# 10. Define inertia in the context of K-Means.

In the context of **K-Means**, **inertia** is the measure of how well the clusters fit the data.

It is defined as the **sum of squared distances** between each data point and the centroid of the cluster it belongs to.

* **Low inertia** → points are close to their centroids → better clustering.
* **High inertia** → points are far from centroids → poorer clustering.

Inertia is commonly used to evaluate clustering quality and to choose the optimal number of clusters using the **Elbow Method**.


## 11. What is the elbow method in K-Means clustering?

The **elbow method** is a technique used in **K-Means clustering** to determine the optimal number of clusters (**K**).

* You run K-Means with different values of K.
* For each K, you calculate the **inertia** (sum of squared distances of points to their cluster centroids).
* As K increases, inertia decreases (clusters fit better).
* You plot **inertia vs. K** and look for a point where the rate of decrease sharply slows down, forming an **“elbow” shape**.
* That K value at the elbow is chosen as the optimal number of clusters.

In short: the elbow method helps pick the best K by balancing clustering accuracy and model simplicity.

## 12. Describe the concept of "density" in DBSCAN.

In **DBSCAN**, "density" refers to how closely data points are packed together in a region.

* A region is considered **dense** if a point has at least **MinPts** neighbors within a radius **ε**.
* Such points are called **core points**, and they form the basis of clusters.
* Areas with low density (fewer than MinPts points within ε) are treated as noise or outliers.

In short: **density in DBSCAN means the concentration of points within a given distance ε, and clusters are formed around dense regions.**

## 13. Can hierarchical clustering be used on categorical data?

Yes, hierarchical clustering can be used on **categorical data**, but with some adjustments:

* The usual **Euclidean distance** isn’t meaningful for categorical features.
* Instead, you use other distance measures such as:

  * **Hamming distance** (proportion of mismatched attributes)
  * **Jaccard similarity** (for binary data)
  * **Gower distance** (for mixed numerical + categorical data)

Once you define an appropriate distance metric, hierarchical clustering works the same way: building a dendrogram and merging/splitting clusters.

In short: **Yes, it can be applied, but you must use a suitable distance metric for categorical data instead of Euclidean distance.**

## 14. What does a negative Silhouette Score indicate?

A **negative Silhouette Score** indicates that, on average, data points are **closer to points in another cluster than to points in their own cluster**.

This usually means:

* The clusters overlap heavily.
* Some points may be **misclassified** into the wrong cluster.
* The chosen number of clusters (K) or clustering method may not fit the data well.

In short: a negative Silhouette Score signals **poor clustering quality** and possible misassignment of points.

## 15.  Explain the term "linkage criteria" in hierarchical clustering.

In **hierarchical clustering**, the **linkage criteria** define how the distance between two clusters is calculated when deciding which clusters to merge.

Common linkage methods are:

* **Single linkage:** distance between the closest pair of points (can cause “chaining”).
* **Complete linkage:** distance between the farthest pair of points (produces compact clusters).
* **Average linkage:** average distance between all pairs of points across clusters.
* **Ward’s method:** minimizes the increase in total variance when clusters are merged.

In short: **linkage criteria determine the rule for measuring cluster-to-cluster distance in hierarchical clustering.**

## 16. Why might K-Means clustering perform poorly on data with varying cluster sizes or densities?

K-Means clustering can perform poorly on data with **varying cluster sizes or densities** because of how it works:

1. **Equal-size assumption** – K-Means assumes clusters are roughly spherical and similar in size. If clusters are very small and large, the centroids may shift toward larger clusters and misclassify smaller ones.

2. **Density sensitivity** – It assumes uniform density. If one cluster is dense and another is sparse, K-Means may incorrectly split or merge clusters.

3. **Centroid bias** – Centroids are based on the mean. Sparse or small clusters can get “pulled” toward larger, denser clusters.

In short: **K-Means struggles with clusters of different sizes and densities because it assumes spherical, evenly sized clusters.**

## 17. What are the core parameters in DBSCAN, and how do they influence clustering?

The two **core parameters** in **DBSCAN** are:

1. **ε (Epsilon):**

   * Defines the **radius** of the neighborhood around a point.
   * A point is considered a neighbor if it lies within ε distance.
   * **Small ε** → many points may be treated as noise.
   * **Large ε** → clusters may merge together.

2. **MinPts (Minimum Points):**

   * The **minimum number of points** (including the point itself) required to form a **dense region** (core point).
   * **Small MinPts** → even sparse regions can form clusters (risk of noise being included).
   * **Large MinPts** → only very dense regions form clusters (risk of missing smaller clusters).

## 18. How does K-Means++ improve upon standard K-Means initialization?

**K-Means++** improves standard K-Means by providing a **smarter initialization of centroids**, which helps avoid poor clustering results.

### Standard K-Means issue:

* Centroids are chosen randomly.
* Bad initial choices can lead to slow convergence or suboptimal clusters.

### K-Means++ solution:

1. Choose the first centroid randomly from the data points.
2. For each remaining point, compute its distance to the nearest chosen centroid.
3. Select the next centroid with a probability proportional to the **square of this distance** (points farther away are more likely to be chosen).
4. Repeat until K centroids are chosen, then proceed with normal K-Means.

### Benefits:

* Centroids start spread out, reducing chances of poor clustering.
* Leads to **faster convergence**.
* Produces **better clustering quality** (lower inertia).

In short: **K-Means++ improves clustering by choosing initial centroids more strategically instead of purely random selection.**

## 19. What is agglomerative clustering?

**Agglomerative clustering** is a type of **hierarchical clustering** that follows a **bottom-up approach**:

1. Start with each data point as its **own cluster**.
2. Iteratively **merge the two closest clusters** based on a chosen distance metric and linkage criterion.
3. Repeat until all points belong to a single cluster or a stopping condition (like a desired number of clusters) is reached.

* The process can be visualized using a **dendrogram**.
* It is useful for discovering hierarchical relationships in data.

In short: **Agglomerative clustering builds clusters by progressively merging the closest points/clusters from the bottom up.**

## 20. What makes Silhouette Score a better metric than just inertia for model evaluation?

The **Silhouette Score** is often considered better than just **inertia** because it evaluates **both cohesion and separation** of clusters, not just compactness:

1. **Inertia** measures only how close points are to their cluster centroids (cohesion).

   * Low inertia can be misleading if clusters are overlapping or poorly separated.

2. **Silhouette Score** considers:

   * **Cohesion**: how close a point is to other points in its own cluster.
   * **Separation**: how far a point is from points in the nearest other cluster.

3. **Interpretation**:

   * Range: -1 to +1.
   * Positive values → well-clustered.
   * Near 0 → points on the boundary.
   * Negative → misassigned points.

In short: **Silhouette Score gives a more holistic measure of clustering quality by accounting for both intra-cluster similarity and inter-cluster separation, whereas inertia only measures intra-cluster compactness.**

```python
# 22.  Load the Iris dataset and use Agglomerative Clustering to group the data into 3 clusters. Display the first 10 predicted labels

from sklearn.cluster import AgglomerativeClustering

data = load_iris()
X = data.data

cluster = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
labels_ = cluster.fit_predict(X)

print("First 10 predicted labels:", labels_[:10])

plt.figure(figsize=(10, 7))
plt.scatter(X[:, 0], X[:, 1], c=labels_, cmap='rainbow', edgecolor='k', s=100)
plt.xlabel(data.feature_names[0])
plt.ylabel(data.feature_names[1])
plt.title("Agglomerative Clustering of Iris Dataset")
plt.show()
```

```python
# 23. Generate synthetic data using make_moons and apply DBSCAN. Highlight outliers in the plot.

from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

X, y = make_moons(n_samples=100, noise=0.05, random_state=42)

dbscan = DBSCAN(eps=0.2, min_samples=5) 
clusters = dbscan.fit_predict(X)
outliers = clusters == -1

plt.figure(figsize=(8, 6))
plt.scatter(X[~outliers, 0], X[~outliers, 1], c=clusters[~outliers], cmap='viridis', s=100, label='Cluster')
plt.scatter(X[outliers, 0], X[outliers, 1], c='red', s=100, label='Outliers', marker='x')
plt.title("DBSCAN Clustering on Moon-Shaped Data")
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.legend()
plt.show()
```

```python
# 24. Load the Wine dataset and apply K-Means clustering after standardizing the features. Print the size of each cluster

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

data = load_wine()
X = data.data
y = data.target 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

n_clusters = 3  
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

unique, counts = pd.Series(clusters).value_counts().sort_index(), pd.Series(clusters).value_counts().sort_index().values
print("Cluster sizes:")
for i, count in enumerate(counts):
    print(f"Cluster {i}: {count} samples")
```
