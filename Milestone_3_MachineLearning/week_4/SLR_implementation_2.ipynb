```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('height-weight.csv')
```

```python
df
```

```python
plt.scatter(df.Weight, df.Height)
plt.xlabel("Weight")
plt.ylabel("Height")
plt.show()
```

```python
# Problem statement >> Predict height based on weight
```

```python
# divide the data into x and y
X = df[['Weight']]
y = df['Height']
```

```python
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

```python
# Scaling (Optional) >> Standardization 
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
# Scaling is optional >> It doesn't affect the distribution of the data
```

```python
plt.scatter(X_train, y_train)
```

```python
# Model training
from sklearn.linear_model import LinearRegression
```

```python
# linear regression paraeters
# fit_intercept >> The best fit line will have intercept, by default it is true
# copy_X >> copy the original X_train and then build the model, It will not modify the original data
# n_jobs >> -1 means using all processors
# positive >> when set o true, forces the coefficient to be positive

# fit >> train the model
# predict >> tets the model
```

```python
regressor =  LinearRegression()
```

```python
regressor
```

```python
regressor.fit(X_train, y_train)
```

```python
regressor.coef_
```

```python
regressor.intercept_
```

```python
print("The Slope or coefficient of model is ", regressor.coef_)
print("The intercept of the model is", regressor.intercept_)
```
