```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings 
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('mpg')
```

```python
df.head()
```

```python
df.drop("name", axis=1, inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.horsepower.median()
```

```python
df['horsepower'] = df['horsepower'].fillna(df.horsepower.median())
```

```python
df.isna().sum()
```

```python
df['origin'].value_counts()
```

```python
df['origin'] = df['origin'].map({"usa": 1, "japan": 2, "europe": 3})
```

```python
df['origin'].value_counts()
```

```python
df.dtypes
```

```python
# seprate into X and y
```

```python
X = df.drop('mpg', axis=1)
y = df['mpg']
```

```python
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
X_test
```

```python
from sklearn.linear_model import LinearRegression

regression_model = LinearRegression()
```

```python
regression_model
```

```python
regression_model.fit(X_train, y_train)
```

```python
regression_model.coef_
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {regression_model.coef_[i]}")
```

```python
from sklearn.metrics import r2_score

y_pred_linear = regression_model.predict(X_test)
r2_linear = r2_score(y_test, y_pred_linear)
r2_linear
```

```python
# regularised model
# ridge regression
```

```python
from sklearn.linear_model import Ridge

ridge_regression_model = Ridge(alpha=0.1)
ridge_regression_model
```

```python
ridge_regression_model.fit(X_train, y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {ridge_regression_model.coef_[i]}")
```

```python
y_pred_ridge = ridge_regression_model.predict(X_test)
r2_ridge = r2_score(y_test, y_pred_ridge)
r2_ridge
```

```python
from sklearn.linear_model import Lasso

lasso_regression_model = Lasso(alpha=0.5)
```

```python
lasso_regression_model
```

```python
lasso_regression_model.fit(X_train, y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {lasso_regression_model.coef_[i]}")
```

```python
y_pred_lasso = lasso_regression_model.predict(X_test)
r2_lasso = r2_score(y_test, y_pred_lasso)
r2_lasso
```

```python
from sklearn.linear_model import ElasticNet

elasticnet_regression_model = ElasticNet(alpha=1, l1_ratio=0.5)
elasticnet_regression_model
```

```python
elasticnet_regression_model.fit(X_train, y_train)
```

```python
for i, col_name in enumerate(X_train.columns):
    print(f"The coefficient for {col_name} is {elasticnet_regression_model.coef_[i]}")
```

```python
y_pred_elasticnet = elasticnet_regression_model.predict(X_test)
r2_elasticnet = r2_score(y_test, y_pred_elasticnet)
r2_elasticnet
```

```python
# regularisation with crossvalidation
from sklearn.linear_model import LassoCV

lasso_cv = LassoCV(cv = 5, verbose=2)
```

```python
lasso_cv.fit(X_train, y_train)
```

```python
y_pred = lasso_cv.predict(X_test)
r2_score(y_test, y_pred)
```

```python
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(cv=5)
ridge_cv.fit(X_train, y_train)
```

```python
y_pred = ridge_cv.predict(X_test)
r2_score(y_test, y_pred)
```
