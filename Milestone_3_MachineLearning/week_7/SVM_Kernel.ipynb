```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings('ignore')
```

```python
x = np.linspace(-6, 6, 100)
```

```python
x
```

```python
y = np.sqrt(10**2 - x**2)
```

```python
y
```

```python
y = np.hstack([y, -y])
x = np.hstack([x, -x])
```

```python
x
```

```python
y
```

```python
x1 = np.linspace(-6, 6, 100)
y1 = np.sqrt(4**2 - x1**2)
y1 = np.hstack([y1, -y1])
x1 = np.hstack([x1, -x1])
```

```python
plt.scatter(y, x)
plt.scatter(y1, x1)
```

```python
df1 = pd.DataFrame(np.vstack([y, x]).T, columns = ["X1", "X2"])
df1["Y"] = 0
```

```python
df1
```

```python
df2 = pd.DataFrame(np.vstack([y1, x1]).T, columns=["X1", "X2"])
df2['Y'] = 1

df = pd.concat([df1, df2], ignore_index=True)
df.head()
```
