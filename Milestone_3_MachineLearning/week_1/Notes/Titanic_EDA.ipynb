```python
# Ways to deal with missing data
# 1. Deleting the column with mmissing data

# No doubt it is one of the quickest techniques one can use to deal with
# missing data but we also have to keep in mind that there is loss of 
# information. This technique should only be used when majority of the 
# values in a column has NULL values.
```

```python
df.isna().sum()
```

```python
df = df.drop(['Cabin'], axis=1)
df.columns
```

```python
df.isnull().sum()
```

```python
# 2. Deleting the row with missing data

# In this method we are deleting rows which has at least one NULL value. 
# This is not the best practice because of the fact that data is information. 
# Even though other values are non null we delete the entire row if there is at 
# least one NULL value. For instance, if every row has some (column) value 
# missing, you might end up deleting the whole data.
```

```python
# Deletes the rows which has atleast one null value

updated_df = df.dropna(axis=0)
```

```python
df.shape
```

```python
updated_df.shape
```

```python
# 3. Imputing missing values with mean/median
# Can not be applied categorical data
# Mean imputation is sensitive to outliers
```

```python
# Mean Imputation
df['Age'] = df['Age'].fillna(df['Age'].mean(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
# We can fill the missing values using group level statistics in the following manner.
df['Age'] = df['Age'].fillna(df.groupby('Sex')['Age'].transform('mean'))
```
